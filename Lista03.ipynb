Questão 1 (exemplo resolvido)

```python
notas = []

for i in range(5):
    nota = float(input("Digite a nota: "))
    notas.append(nota)

soma = 0

for nota in notas:
    soma += nota

media = soma / len(notas)

acima_ou_igual = 0

for nota in notas:
    if nota >= media:
        acima_ou_igual += 1

print("Notas:", notas)
print("Média:", media)
print("Notas >= média:", acima_ou_igual)
```

**Questão 4**

Leia 10 números inteiros e armazene todos em uma lista. Depois, percorra a lista e crie duas novas listas: uma apenas com os pares e outra apenas com os ímpares. Mostre as três listas.

```python
numeros = []
pares = []
impares = []

for i in range(10):
    numero = int(input("Digite um número inteiro: "))
    numeros.append(numero)

for numero in numeros:
    if numero % 2 == 0:
        pares.append(numero)
    else:
        impares.append(numero)

print("Lista original:", numeros)
print("Pares:", pares)
print("Ímpares:", impares)
```

**Questão 8**

Leia 6 números inteiros, forme uma tupla com eles e informe: quantas vezes o número 5 apareceu, a posição da primeira ocorrência de 5 (se existir) e quantos valores são pares.

```python
valores = []

for i in range(6):
    numero = int(input("Digite um número inteiro: "))
    valores.append(numero)

numeros = tuple(valores)

quantidade_5 = 0
pares = 0
posicao_5 = -1

for i in range(len(numeros)):
    if numeros[i] == 5:
        quantidade_5 += 1

        if posicao_5 == -1:
            posicao_5 = i

    if numeros[i] % 2 == 0:
        pares += 1

print("Tupla:", numeros)
print("Quantidade de vezes que o 5 apareceu:", quantidade_5)

if posicao_5 != -1:
    print("Posição da primeira ocorrência do 5:", posicao_5)
else:
    print("O número 5 não apareceu.")

print("Quantidade de valores pares:", pares)
```

7

Tupla: (5, 8, 3, 5, 10, 7)

Quantidade de valores pares: 2

**Questão 12**

Crie um dicionário com 4 produtos e suas quantidades em estoque. Leia o nome de um produto e uma quantidade vendida. Se o produto existir e houver estoque suficiente, atualize o dicionário; caso contrário, informe o problema. Mostre o estoque final.

```python
estoque = {
    "arroz": 10,
    "feijão": 15,
    "macarrão": 20,
    "café": 8
}

produto = input("Digite o nome do produto: ")
quantidade = int(input("Digite a quantidade vendida: "))

if produto not in estoque:
    print("Produto não cadastrado.")
elif quantidade > estoque[produto]:
    print("Estoque insuficiente.")
else:
    estoque[produto] -= quantidade
    print("Venda realizada com sucesso.")

print("Estoque final:", estoque)
```

Entrada:

3

Saída:
Venda realizada com sucesso.
Estoque final: {'arroz': 7, 'feijão': 15, 'macarrão': 20, 'café': 8}

**Questão 15**

Construa um pequeno cadastro de 5 itens de estoque. Cada item deve ser um dicionário com código, nome, quantidade e preço. Armazene os dicionários em uma lista. Depois: (a) mostre todos os itens; (b) calcule o valor total do estoque; (c) identifique o item de maior valor em estoque (quantidade × preço); (d) crie uma tupla resumo contendo (quantidade_de_itens_cadastrados, valor_total_estoque).

```python
estoque = [
    {
        "codigo": 1,
        "nome": "Arroz",
        "quantidade": 10,
        "preco": 25.00
    },
    {
        "codigo": 2,
        "nome": "Feijão",
        "quantidade": 15,
        "preco": 8.50
    },
    {
        "codigo": 3,
        "nome": "Café",
        "quantidade": 8,
        "preco": 15.00
    },
    {
        "codigo": 4,
        "nome": "Leite",
        "quantidade": 20,
        "preco": 6.50
    },
    {
        "codigo": 5,
        "nome": "Macarrão",
        "quantidade": 12,
        "preco": 5.00
    }
]

valor_total = 0
maior_valor = 0
item_maior = None

for item in estoque:
    valor_item = item["quantidade"] * item["preco"]
    valor_total += valor_item

    if valor_item > maior_valor:
        maior_valor = valor_item
        item_maior = item

print("Itens do estoque:")

for item in estoque:
    print(item)

print("Valor total do estoque:", valor_total)

print("Item de maior valor em estoque:")
print(item_maior)
print("Valor:", maior_valor)

resumo = (len(estoque), valor_total)

print("Tupla resumo:", resumo)
```
