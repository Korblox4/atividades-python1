1. Maior e menor entre vários valores

1: Ler 6 números inteiros e informar qual é o maior e qual é o menor valor digitado.

2: Entradas: 6 números inteiros; Saída: maior e menor valor; Informações guardadas: primeiro, numero, maior e menor.

3: Repetição: sim, para ler os 6 números; Atualização: maior e menor; Classificação: sim, comparando cada número com o maior e menor atuais; Condição de parada: após ler os 6 números.

```python
primeiro = int(input("Digite um número: "))

maior = primeiro
menor = primeiro

for i in range(5):
    numero = int(input("Digite um número: "))

    if numero > maior:
        maior = numero

    if numero < menor:
        menor = numero

print("Maior:", maior)
print("Menor:", menor)
```

2. Estatísticas de uma turma

1: Ler as notas de 5 alunos, calcular a média da turma e informar quantos alunos tiveram nota maior ou igual a 7 e quantos tiveram nota menor que 7.

2: Entradas: 5 notas; Saída: média, quantidade de notas maiores ou iguais a 7 e quantidade de notas menores que 7; Informações guardadas: nota, soma, aprovados e abaixo.

3: Repetição: sim, para ler as 5 notas; Atualização: soma, aprovados e abaixo; Classificação: sim, para separar as notas maiores ou iguais a 7 das menores que 7; Condição de parada: após ler as 5 notas.

```python
soma = 0
aprovados = 0
abaixo = 0

for i in range(5):
    nota = float(input("Nota: "))

    soma += nota

    if nota >= 7:
        aprovados += 1
    else:
        abaixo += 1

media = soma / 5

print("Média:", media)
print("Notas >= 7:", aprovados)
print("Notas < 7:", abaixo)
```

3. Validando uma entrada

1: Pedir uma nota entre 0 e 10 e continuar solicitando uma nova nota enquanto o valor informado estiver fora desse intervalo.

2: Entrada: uma nota; Saída: a nota válida; Informações guardadas: nota.

3: Repetição: sim, enquanto a nota for inválida; Atualização: nota recebe um novo valor; Classificação: sim, para verificar se a nota é válida ou inválida; Condição de parada: quando a nota estiver entre 0 e 10.

```python
nota = float(input("Nota entre 0 e 10: "))

while nota < 0 or nota > 10:
    print("Nota inválida.")
    nota = float(input("Digite novamente: "))

print("Nota válida:", nota)
```

4. Estacionamento

1: Ler a quantidade inteira de horas que o carro ficou no estacionamento e calcular o valor a pagar de acordo com a faixa de cobrança.

2: Entrada: quantidade de horas; Saída: valor a pagar; Informações guardadas: horas e valor.

3: Repetição: não; Atualização: não; Classificação: sim, para identificar a faixa de cobrança; Condição de parada: não.

```python
horas = int(input("Digite a quantidade de horas: "))

if horas <= 2:
    valor = 8

elif horas <= 5:
    valor = 8 + (horas - 2) * 3

else:
    valor = 17 + (horas - 5) * 2

print("Valor a pagar: R$", valor)
```

5. Máquina de venda de ingresso

1: Ler valores inteiros até que o total inserido seja suficiente para comprar um ingresso de R$ 25,00 e, depois, informar o total inserido e o troco.

2: Entrada: valores inteiros inseridos; Saída: total inserido e troco; Informações guardadas: preco, total e valor.

3: Repetição: sim, até que o total seja suficiente; Atualização: total recebe os valores inseridos; Classificação: não; Condição de parada: quando total for maior ou igual a 25.

```python
preco = 25
total = 0

while total < preco:
    valor = int(input("Insira um valor: "))
    total += valor

troco = total - preco

print("Total inserido:", total)
print("Troco:", troco)
```

6. Maior valor e sua posição

1: Ler 8 números inteiros e informar o maior valor e a posição em que ele apareceu pela primeira vez.

```python
maior = int(input("Digite um número: "))
posicao_maior = 1

for posicao in range(2, 9):
    numero = int(input("Digite um número: "))

    if numero > maior:
        maior = numero
        posicao_maior = posicao

print("Maior:", maior)
print("Primeira posição:", posicao_maior)
```

7. Segundo maior valor

1: Ler 8 números inteiros distintos e informar qual é o maior e qual é o segundo maior valor.

2: Entrada: 8 números inteiros distintos; Saída: maior e segundo maior; Informações guardadas: maior, segundo_maior, primeiro, segundo e numero.

3: Repetição: sim, para ler os 8 números; Atualização: maior e segundo_maior; Classificação: sim, para verificar se o número é maior que o maior ou que o segundo maior; Condição de parada: após ler os 8 números.

```python
primeiro = int(input("Digite um número: "))
segundo = int(input("Digite um número: "))

if primeiro > segundo:
    maior = primeiro
    segundo_maior = segundo
else:
    maior = segundo
    segundo_maior = primeiro

for i in range(6):
    numero = int(input("Digite um número: "))

    if numero > maior:
        segundo_maior = maior
        maior = numero

    elif numero > segundo_maior:
        segundo_maior = numero

print("Maior:", maior)
print("Segundo maior:", segundo_maior)
```

8. Detectando repetição consecutiva

1: Ler 8 números inteiros e verificar se em algum momento dois números iguais foram digitados consecutivamente.

2: Entrada: 8 números inteiros; Saída: se existem ou não valores consecutivos iguais; Informações guardadas: anterior, numero e repetiu.

3: Repetição: sim, para ler os números; Atualização: anterior recebe o número atual; Classificação: sim, comparando o número atual com o anterior; Condição de parada: após ler os 8 números.

```python
anterior = int(input("Digite um número: "))
repetiu = False

for i in range(7):
    numero = int(input("Digite um número: "))

    if numero == anterior:
        repetiu = True

    anterior = numero

if repetiu:
    print("Existem valores consecutivos iguais.")
else:
    print("Não existem valores consecutivos iguais.")
```

9. Controle de saldo

1: Começar com R$ 500,00, realizar 6 operações de depósito ou saque e, no final, informar o saldo. Um saque só pode ser realizado quando houver saldo suficiente.

2: Entradas: tipo e valor de cada operação; Saída: mensagens de saldo insuficiente e saldo final; Informações guardadas: saldo, tipo e valor.

3: Repetição: sim, para realizar 6 operações; Atualização: saldo é alterado conforme a operação; Classificação: sim, para identificar depósito ou saque; Condição de parada: após 6 operações.

```python
saldo = 500

for i in range(6):
    tipo = int(input("Tipo (1 = depósito, 2 = saque): "))
    valor = float(input("Valor: "))

    if tipo == 1:
        saldo += valor

    elif tipo == 2:
        if valor <= saldo:
            saldo -= valor
        else:
            print("Saldo insuficiente.")

print("Saldo final:", saldo)
```

11. Número primo

1: Ler um número inteiro maior que 1 e verificar se ele é primo.

2: Entrada: um número inteiro maior que 1; Saída: se o número é primo ou não; Informações guardadas: numero, i e primo.

3: Repetição: sim, para procurar divisores; Atualização: primo pode mudar para False; Classificação: sim, verificando se existe algum divisor; Condição de parada: após testar os possíveis divisores.

```python
numero = int(input("Digite um número maior que 1: "))

primo = True

for i in range(2, numero):
    if numero % i == 0:
        primo = False

if primo:
    print("É primo.")
else:
    print("Não é primo.")
```

12. Divisores de um número

1: Ler um número inteiro positivo, mostrar todos os seus divisores positivos e informar a quantidade de divisores encontrados.

2: Entrada: número inteiro positivo; Saída: divisores e quantidade de divisores; Informações guardadas: n, i e quantidade.

3: Repetição: sim, para testar os números de 1 até n; Atualização: quantidade aumenta quando um divisor é encontrado; Classificação: sim, para verificar se cada número é divisor; Condição de parada: após testar todos os números até n.

```python
n = int(input("Digite um número positivo: "))

quantidade = 0

for i in range(1, n + 1):
    if n % i == 0:
        print(i)
        quantidade += 1

print("Quantidade de divisores:", quantidade)
```

13. Números primos em um intervalo

1: Ler dois números positivos A e B, com A menor que B, mostrar os números primos existentes entre eles e informar quantos foram encontrados.

2: Entradas: A e B; Saída: números primos e quantidade encontrada; Informações guardadas: a, b, numero, i, primo e quantidade.

3: Repetição: sim, para percorrer o intervalo e testar os divisores; Atualização: primo e quantidade; Classificação: sim, para verificar se cada número é primo; Condição de parada: quando todos os números de A até B forem analisados.

```python
a = int(input("Digite A: "))
b = int(input("Digite B: "))

quantidade = 0

for numero in range(a, b + 1):

    if numero > 1:
        primo = True

        for i in range(2, numero):
            if numero % i == 0:
                primo = False

        if primo:
            print(numero)
            quantidade += 1

print("Quantidade de primos:", quantidade)
```

14. Jogo com limite de tentativas

1: Usar o número secreto 37 e permitir que o usuário tente acertá-lo no máximo 5 vezes, informando se o número secreto é maior ou menor que o palpite após cada erro.

2: Entrada: palpites do usuário; Saída: mensagens indicando se o secreto é maior, menor ou se o usuário acertou; Informações guardadas: secreto, tentativas, palpite e acertou.

3: Repetição: sim, até 5 tentativas ou até acertar; Atualização: tentativas aumenta a cada palpite e acertou muda quando o número é encontrado; Classificação: sim, comparando o palpite com o número secreto; Condição de parada: acertar o número ou chegar a 5 tentativas.

```python
secreto = 37
tentativas = 0
acertou = False

while tentativas < 5 and not acertou:
    palpite = int(input("Digite seu palpite: "))
    tentativas += 1

    if palpite == secreto:
        acertou = True
        print("Você acertou!")

    elif palpite < secreto:
        print("O número secreto é maior.")

    else:
        print("O número secreto é menor.")

if not acertou:
    print("Fim das tentativas.")
```

15. Desafio final — Caixa de atendimento

1: Iniciar o caixa com R$ 200,00, receber operações de entrada ou retirada até que o código 0 seja informado e controlar o saldo, as operações realizadas e os valores movimentados. Retiradas sem saldo suficiente devem ser recusadas.

2: Entradas: código e valor da operação; Saída: saldo final, quantidade de entradas, quantidade de retiradas realizadas, quantidade de retiradas recusadas, total que entrou e total que saiu; Informações guardadas: saldo, codigo, valor, entradas, retiradas, recusadas, total_entradas e total_saidas.

3: Repetição: sim, até o código 0; Atualização: saldo, contadores e totais; Classificação: sim, para identificar entrada, retirada ou encerramento; Condição de parada: quando o código informado for 0.

```python
saldo = 200

entradas = 0
retiradas = 0
recusadas = 0

total_entradas = 0
total_saidas = 0

codigo = int(input("Código (1 = entrada, 2 = retirada, 0 = sair): "))

while codigo != 0:

    if codigo == 1:
        valor = float(input("Valor da entrada: "))

        saldo += valor
        entradas += 1
        total_entradas += valor

    elif codigo == 2:
        valor = float(input("Valor da retirada: "))

        if valor <= saldo:
            saldo -= valor
            retiradas += 1
            total_saidas += valor
        else:
            print("Saldo insuficiente.")
            recusadas += 1

    else:
        print("Código inválido.")

    codigo = int(input("Código (1 = entrada, 2 = retirada, 0 = sair): "))

print("Saldo final:", saldo)
print("Quantidade de entradas:", entradas)
print("Quantidade de retiradas realizadas:", retiradas)
print("Quantidade de retiradas recusadas:", recusadas)
print("Total que entrou:", total_entradas)
print("Total que saiu:", total_saidas)
```
