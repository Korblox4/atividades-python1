Questão 1

Faça um programa que leia um número inteiro e informe se ele é positivo, negativo ou igual a zero.

Questão 5

Leia duas notas de um aluno e calcule sua média. Informe se ele foi aprovado, está em recuperação ou foi reprovado. Ao final, mostre também a média calculada.

```python
nota1 = float(input("Digite a primeira nota: "))
nota2 = float(input("Digite a segunda nota: "))

media = (nota1 + nota2) / 2

if media >= 7:
    print("Aprovado")
elif media >= 5:
    print("Recuperação")
else:
    print("Reprovado")

print("Média:", media)
```

Questão 13

Considere que a senha correta do sistema é 1234. Peça ao usuário que digite a senha. Enquanto a senha estiver incorreta, solicite uma nova tentativa. Quando acertar, mostre "Acesso permitido!".

```python
senha_correta = 1234

senha = int(input("Digite a senha: "))

while senha != senha_correta:
    senha = int(input("Senha incorreta. Digite novamente: "))

print("Acesso permitido!")
```

Questão 14

Considere que o número secreto é 7. Peça ao usuário que tente descobrir o número. Enquanto ele não acertar, informe se deve tentar um número maior ou menor. Ao acertar, mostre "Parabéns! Você acertou." e informe quantas tentativas foram necessárias.

```python
numero_secreto = 7
tentativas = 0

numero = int(input("Digite um número: "))
tentativas = tentativas + 1

while numero != numero_secreto:
    if numero < numero_secreto:
        print("Tente um número maior")
    else:
        print("Tente um número menor")

    numero = int(input("Digite um número: "))
    tentativas = tentativas + 1

print("Parabéns! Você acertou.")
print("Tentativas:", tentativas)
```

Questão 15

Leia números inteiros repetidamente. O programa deve parar quando o usuário digitar 0. Ao final, informe a quantidade de números digitados, a quantidade de positivos, a quantidade de negativos e a soma de todos os números digitados.

```python
quantidade = 0
positivos = 0
negativos = 0
soma = 0

numero = int(input("Digite um número: "))

while numero != 0:
    quantidade = quantidade + 1
    soma = soma + numero

    if numero > 0:
        positivos = positivos + 1
    else:
        negativos = negativos + 1

    numero = int(input("Digite um número: "))

print("Quantidade:", quantidade)
print("Positivos:", positivos)
print("Negativos:", negativos)
print("Soma:", soma)
```
